```python
from sklearn.cluster import KMeans
from sklearn.preprocessing import StandardScaler
import matplotlib.pyplot as plt 
import pandas as pd 
```

```python
df = pd.read_csv("Data/ignore/panel_data.csv")
df.head()
```

```python
plt.scatter(df["distance"], df["reg_delt"])
plt.xlabel("Distance to hurricane")
plt.ylabel("Gas price (reg delt)")
plt.title("Distance and change in gas prices")
plt.show()
```

## Cluster 

```python
df = df.dropna(subset=["reg_delt", "distance"])

X = df[["distance", "reg_delt"]]

X.isna().sum()
```

```python
scaler = StandardScaler()

X_scaled = scaler.fit_transform(X)

kmeans = KMeans(n_clusters =3, random_state=0)
df["cluster"] = kmeans.fit_predict(X_scaled)
```

```python
plt.scatter(df["distance"], df["reg_delt"], c=df["cluster"])
plt.xlabel("Distance to hurricane")
plt.ylabel("Gas price (reg delt)")
plt.title("Clusters (3)")
plt.show()
```
