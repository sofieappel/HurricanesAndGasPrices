```python
import pandas as pd
import matplotlib.pyplot as plt
import numpy as np
import statsmodels.formula.api as smf 
```

## data load in

```python
path = "Data/gas_coordinates.csv"
gas = pd.read_csv(path)
gas["date"] = pd.to_datetime(gas["date"])
gas["month"] = gas["date"].dt.month

gas.head()
```

```python
path = "Data/ignore/hurr_mod.csv"
hurricanes = pd.read_csv(path)
hurricanes["date"] = pd.to_datetime(hurricanes["DATE"])
```

## combining

```python
hurricanes = hurricanes.drop_duplicates(subset = "date", keep = "first")
hurricanes.shape
```

```python
gas.shape
```

```python
panel = gas.merge(hurricanes, on = "date", how = "left")
panel.shape
```

```python
print("gas:", gas.shape)
print("hurricanes:", hurricanes.shape)
print("panel:", panel.shape)

panel[["date", "met", "lat", "lng", "USA_LAT", "USA_LON"]].head(10)
```

```python
panel["distance"] = np.sqrt(
    (panel["lat"] - panel['USA_LAT'])**2 + 
    (panel["lng"] - panel['USA_LON'])**2
)
```
